Este programa crea una serie de datos original con 15 valores aleatorios entre 1 y 3. Luego, calcula los conteos de frecuencia de cada valor único utilizando el método value_counts. Obtiene el valor más frecuente con idxmax y reemplaza todos los demás valores con 'Otro' utilizando el método apply. Finalmente, imprime la serie original, las frecuencias principales, y la serie modificada.

```python
# Crear la serie de datos original con valores aleatorios entre 1 y 3
data = pd.Series(np.random.randint(1, 4, 15))

print("Serie original:")
print(data)
print("tipo:", data.dtype)

# Calcular los conteos de frecuencia de cada valor único
frequency_counts = data.value_counts()

print("\nFrecuencias principales:")
print(frequency_counts)
print("tipo:", frequency_counts.dtype)

# Obtener el valor más frecuente
most_frequent_value = frequency_counts.idxmax()

# Reemplazar todos los valores excepto el más frecuente con 'Otro'
data_replaced = data.apply(lambda x: x if x == most_frequent_value else 'Otro')

print("\nSerie después de reemplazar todo lo demás con 'Otro':")
print(data_replaced)
print("tipo:", data_replaced.dtype)
```

Ej 2 Escribe un programa Pandas para conectarte a la base de datos datafundamentals y guardar el df de pagos en una tabla nueva. Luego muestra las tablas en la bbdd y comprueba que existe la nueva tabla.

```python
import pandas as pd
import psycopg2
from sqlalchemy import create_engine

# Datos de conexión
PGHOST = 'ep-autumn-violet-a5sfis6z.us-east-2.aws.neon.tech'
PGDATABASE = 'data fundamentals'
PGUSER = 'data fundamentals_owner'
PGPASSWORD = 't6v1eQALByal'

# Función para importar datos de la tabla 'payments'
def importar():
    SQL = 'SELECT * FROM payments'
    print('SQL:', SQL)
    
    try:
        conn = psycopg2.connect(
            database=PGDATABASE,
            user=PGUSER,
            password=PGPASSWORD,
            host=PGHOST,
            port=5432
        )
        df = pd.read_sql(SQL, conn)
    except Exception as err:
        print(f"Error: '{err}'")
        df = None
    finally:
        if conn:
            conn.close()
    return df

# Importar datos de la tabla 'payments'
resultado = importar()
print(resultado)

if resultado is not None:
    # Crear una conexión usando SQLAlchemy
    engine = create_engine(f'postgresql+psycopg2://{PGUSER}:{PGPASSWORD}@{PGHOST}/{PGDATABASE}')

    # Guardar el DataFrame en una nueva tabla llamada 'new_payments'
    resultado.to_sql('new_payments', engine, if_exists='replace', index=False)

    # Conectarse a la base de datos para listar las tablas
    conn = psycopg2.connect(
        database=PGDATABASE,
        user=PGUSER,
        password=PGPASSWORD,
        host=PGHOST,
        port=5432
    )

    # Crear un cursor
    cursor = conn.cursor()

    # Ejecutar una consulta para listar las tablas
    cursor.execute("SELECT table_name FROM information_schema.tables WHERE table_schema='public'")

    # Obtener los resultados
    tables = cursor.fetchall()

    # Imprimir las tablas
    print("Tablas en la base de datos:")
    for table in tables:
        print(table[0])

    # Comprobar si la nueva tabla existe
    if ('new_payments',) in tables:
        print("La tabla 'new_payments' ha sido creada exitosamente.")
    else:
        print("La tabla 'new_payments' no se ha creado.")

    # Cerrar la conexión
    cursor.close()
    conn.close()
else:
    print("No se pudo importar el DataFrame.")
```

En tu ejemplo, deseas crear una tabla de contingencia que muestre el promedio de los valores de la columna colesterol para cada combinación de sexo y peso.

```python
ct_means =pd.crosstab(dfcol['sexo'], dfcol['peso'],values=dfcol['colesterol'], aggfunc='mean',margins=True)
ct_means
```
