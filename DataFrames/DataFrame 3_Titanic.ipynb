1. Escribe un programa Pandas para dividir los pasajeros en cuatro categorías según su edad: (0, 10), (10, 30), (30, 60), (60, 80)

¡Claro! Tanto pd.cut como pd.qcut son funciones de pandas utilizadas para segmentar datos en intervalos, pero tienen diferencias importantes:

Criterio de Segmentación:

pd.cut: Divide los datos en intervalos que tienen el mismo rango (es decir, los intervalos son de igual tamaño). Es útil cuando quieres crear intervalos específicos.
pd.qcut: Divide los datos en intervalos que contienen aproximadamente el mismo número de observaciones (es decir, los intervalos tienen la misma cantidad de datos). Es útil para crear cuartiles, deciles, etc.

Parámetros:

pd.cut: Los parámetros principales son x (los datos a segmentar), bins (el número de intervalos o los límites de los intervalos), y labels (para etiquetar los intervalos).
pd.qcut: Los parámetros principales son x (los datos a segmentar), q (el número de cuartiles, deciles, etc.), y labels (para etiquetar los intervalos).

```python
#SE PUEDE HACER CON CUT O CON GROUPBY
df['cage'] = pd.cut(df['age'], [0,  10,  30, 60, 80])
df.head()
```

¡Buena pregunta! Tanto df.pivot como df.pivot_table son funciones de pandas utilizadas para reorganizar datos, pero tienen algunas diferencias clave:

Uso y Flexibilidad:

df.pivot: Es más simple y directo, pero tiene limitaciones. Solo permite pivotar con una columna de valores y no maneja agregaciones.
df.pivot_table: Es más flexible y potente. Permite realizar agregaciones y trabajar con múltiples columnas de valores.
Agregaciones:

df.pivot: No realiza agregaciones. Si hay duplicados en los índices, lanzará un error.
df.pivot_table: Permite especificar una función de agregación (como mean, sum, etc.) para manejar duplicados.
Parámetros:

df.pivot: Utiliza los parámetros index, columns y values.
df.pivot_table: Además de index, columns y values, incluye parámetros adicionales como aggfunc (función de agregación), fill_value (para rellenar valores faltantes), y margins (para agregar totales).

3. Escribe un programa Pandas para encontrar la tasa de supervivencia por género y edad de las diferentes categorías de varias clases. Agregar la tarifa como una dimensión de columnas y dividir la columna de tarifa en 2 categorías según los valores presentes en las columnas de tarifa

pd.qcut(ndf['fare'], 2, labels=['Low Fare', 'High Fare']):

pd.qcut es una función de pandas que divide los datos en intervalos basados en cuantiles. En este caso, se está dividiendo la columna 'fare' del DataFrame ndf en 2 intervalos.
ndf['fare'] es la columna del DataFrame ndf que contiene las tarifas.
2 indica que queremos dividir los datos en 2 intervalos (cuantiles).
labels=['Low Fare', 'High Fare'] asigna etiquetas a los intervalos resultantes. El primer intervalo (el 50% inferior de las tarifas) se etiquetará como 'Low Fare' y el segundo intervalo (el 50% superior de las tarifas) se etiquetará como 'High Fare'.
Asignación a una nueva columna:

ndf['fare_category'] = ... asigna el resultado de pd.qcut a una nueva columna en el DataFrame ndf llamada 'fare_category'.
Esta nueva columna contendrá las etiquetas 'Low Fare' y 'High Fare' para cada fila, dependiendo de en qué intervalo cae la tarifa correspondiente.

```python
# Dividir la columna 'fare' en 2 categorías
ndf['fare_category'] = pd.qcut(ndf['fare'], 2, labels=['Low Fare', 'High Fare'])
ndf['fare_category'] 
```

Utiliza la función pivot_table de pandas para calcular la tasa de supervivencia promedio (columna 'survived') agrupada por género ('sex'), edad ('cage'), clase ('class') y categoría de tarifa ('fare_category'). La función aggfunc='mean' se usa para calcular el promedio de la columna 'survived' para cada grupo.

```python

# Calcular la tasa de supervivencia por género, edad, clase y categoría de tarifa
survival_rate = ndf.pivot_table('survived', index=['sex', 'cage', 'class'], columns='fare_category', aggfunc='mean')

# Mostrar el resultado
print(survival_rate)
```

4. Escribe un programa Pandas para calcular la cantidad de mujeres y hombres que había en cada clase de cabina.

ndf.groupby(['pclass', 'sex']): Agrupa el DataFrame ndf por las columnas 'pclass' y 'sex'.
.size(): Calcula el tamaño de cada grupo, es decir, cuenta el número de ocurrencias en cada grupo.
.unstack(fill_value=0): Convierte el resultado de la agrupación en un DataFrame con un índice jerárquico, donde las filas corresponden a 'pclass' y las columnas a 'sex'. Si hay valores faltantes, los rellena con 0.

```python
# Calcular la cantidad de mujeres y hombres en cada clase de cabina
result = ndf.groupby(['pclass', 'sex']).size().unstack(fill_value=0)

print(result)
```

5. Escribe un programa Pandas para separar el género según si viajaron solos o no y obtener la probabilidad de supervivencia.

```python
# Separar por género y si viajaron solos o no
grouped = ndf.groupby(['sex', 'alone'])

# Calcular la probabilidad de supervivencia
survival_prob = grouped['survived'].mean().reset_index()

# Renombrar la columna para mayor claridad
survival_prob.rename(columns={'survived': 'survival_probability'}, inplace=True)

print(survival_prob)
```

6. Escribe un programa Pandas para crear encontrar la probabilidad de supervivencia por clase, género, embarque en solitario y puerto de embarque.

```python
# Agrupar por 'class', 'sex', 'alone' y 'embarked' y calcular la probabilidad de 'survived'
probabilidad_survived = ndf.groupby(['class', 'sex', 'alone', 'embarked'])['survived'].mean().reset_index()

print(probabilidad_survived)
```
