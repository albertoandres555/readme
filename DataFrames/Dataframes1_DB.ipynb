para el análisis de datos, .loc y .iloc son dos métodos utilizados para acceder a los datos en un DataFrame, pero tienen diferencias clave:

.loc: Se utiliza para acceder a un grupo de filas y columnas por etiquetas o un array booleano. Por ejemplo, df.loc[1:3, 'columna'] selecciona las filas de la 1 a la 3 y la columna llamada 'columna'.

.iloc: Se utiliza para acceder a un grupo de filas y columnas por índices enteros. Por ejemplo, df.iloc[1:3, 0] selecciona las filas de la 1 a la 3 y la primera columna (índice 0).

.loc[]: Es un método de pandas que se utiliza para acceder a un grupo de filas y columnas por etiquetas o una matriz booleana.
(103, 'OM314933'): Es una tupla que especifica la etiqueta de la fila y la etiqueta de la columna que se desea acceder.
En resumen, este código accede al valor en la fila con la etiqueta 103 y la columna con la etiqueta 'OM314933' en el DataFrame midf.

```python
midf.loc[(103,'OM314933')]
```

EJ 8 Escribe un programa pandas que utilice .loc para dividir un DataFrame en función de las etiquetas de filas y columnas.

```python
import pandas as pd

# Crear un DataFrame de ejemplo
data = {
    'A': [1, 2, 3, 4],
    'B': [5, 6, 7, 8],
    'C': [9, 10, 11, 12],
    'D': [13, 14, 15, 16]}

index_labels = ['row1', 'row2', 'row3', 'row4']
df = pd.DataFrame(data, index=index_labels)
df
```

```python
# Dividir el DataFrame utilizando .loc
# Seleccionar filas 'row1' y 'row2' y columnas 'A' y 'B'
df_dividido = df.loc[['row1', 'row2'], ['A', 'B']]

# Mostrar el DataFrame dividido
print("\nDataFrame dividido:")
print(df_dividido)
```
