+Carga el dataset de vuelos "flights14.csv"

```python
import pandas as pd
```

```python
dfligh = pd.read_csv('./flights14.csv',index_col=0)
dfligh.head()
```

escribe un programa Pandas para dividir los datos basados en origen u destino. Muestra cada grupo y el número de elementos por grupo

```python
# Agrupar por 'origen' y 'destino'
grupos = dfligh.groupby(['origin', 'dest'])

# Mostrar cada grupo y el número de elementos por grupo
for (origin, dest), grupo in grupos:
    print(f"Grupo origen={origin}, destino={dest}")
    print(grupo)
    print(f"Número de elementos en el grupo: {len(grupo)}\n")
```

Escribe un programa Pandas para calcula la media, desviación estándar, min y max del retraso total de cada orgige-destino

```python
# Calcular la columna 'total_delay'
dfligh['total_delay'] = dfligh['dep_delay'] + dfligh['arr_delay']

# Agrupar por 'origin' y 'dest' y calcular las estadísticas
stats = dfligh.groupby(['origin', 'dest'])['total_delay'].agg(['mean', 'std', 'min', 'max'])

# Mostrar las estadísticas
print(stats)
```

Escribe un programa Pandas para dividir un conjunto de datos y agruparlos por dos columnas, y luego ordenar los resultados agregados dentro de los grupos.
En el siguiente conjunto de datos de vuelos, agrupar "date" (fecha completa) y "carrier". Luego ordenar la suma de 'total_delay' dentro de los grupos.

```python
dfligh['date'] = pd.to_datetime(dfligh[['year', 'month', 'day']])

# Calcular la columna 'total_delay'
dfligh['total_delay'] = dfligh['dep_delay'] + dfligh['arr_delay']

gdf = dfligh.groupby(['date', 'carrier']).agg({"total_delay": sum})
gdf["total_delay"].groupby(level=0, group_keys=False).nlargest()
```

Escribe un programa Pandas para contar el múmero de retrasos de salida por mes y origen.

dfligh.groupby(['month', 'origin'])['dep_delay'].size() agrupa el DataFrame dfligh por las columnas month y origin, y luego cuenta el número de ocurrencias de dep_delay en cada grupo.
.reset_index(name='count') convierte el resultado en un DataFrame y renombra la columna de conteo a count.
print(count_dep_delay.to_string(index=False)) muestra el DataFrame resultante sin los índices.
Este método es útil para obtener el número de ocurrencias de un valor específico en diferentes grupos dentro de un DataFrame.

```python
# Agrupar por 'month' y 'origin' y contar el número de 'dep_delay'
count_dep_delay = dfligh.groupby(['month', 'origin'])['dep_delay'].size().reset_index(name='count')

# Mostrar los resultados
print(count_dep_delay.to_string(index=False))
```

5. Escribe un programa pandas para cargar los datos de productos y ventas del archivo foodmarkets.xlsx y mezclarlos en un solo dataframe

```python
import pandas as pd

# Cargar las hojas de cálculo del archivo Excel
productos_df = pd.read_excel('./FoodMarket.xlsx', sheet_name='Products')
ventas_df = pd.read_excel('./FoodMarket.xlsx', sheet_name='Purchases')

# Mezclar los DataFrames en uno solo
merged_df = pd.merge(productos_df, ventas_df, on='Product')

# Mostrar las primeras filas del DataFrame resultante
print(merged_df.head())
```

REESTRUCTURAR DATAFRAMES

```python
datos = {'nombre': ['María', 'Luis', 'Carmen'],
         'edad': [18, 22, 20],
         'Matemáticas': [8.5, 7, 3.5],'Economia': [5, 7, 3],'Programación': [8, 7, 5],}
```

```python
df= pd.DataFrame(datos)
print(tabulate(df, headers='keys'))
```

id_vars=['nombre'] especifica que la columna nombre debe mantenerse como identificador.
value_vars=['Matemáticas', 'Economia', 'Programación'] especifica las columnas que deben ser transformadas de formato ancho a formato largo.
var_name='Asignatura' especifica el nombre de la nueva columna que contendrá los nombres de las asignaturas.
value_name='Nota' especifica el nombre de la nueva columna que contendrá las notas correspondientes.

```python
dfm =df.melt(id_vars=['nombre'], value_vars=['Matemáticas', 'Economia', 'Programación'], var_name='Asignatura', value_name='Nota')
dfm
```

columns='Asignatura' especifica que los valores únicos de la columna Asignatura deben convertirse en las nuevas columnas.
index='nombre' especifica que los valores únicos de la columna nombre deben convertirse en el índice del DataFrame.
values='Nota' especifica que los valores de la columna Nota deben llenarse en las celdas del DataFrame.
El DataFrame resultante df_pivot está en formato ancho, donde cada columna representa una asignatura y cada fila representa un estudiante con sus respectivas notas.

```python
dfm.pivot(columns='Asignatura', index='nombre', values='Nota')
```
